3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

1.

```python
result_matrix = np.dot(B,A)
print(result_matrix)
```

2.

```python
result = np.dot(result_matrix,C)
print(result)
```

3.

```python
result_D = np.dot(result_matrix,D)
print(result_D)
```

```python
result_matrix_t = result_matrix.T
print("Транспонована матриця result_matrix", result_matrix_t, sep='\n')
```

2.

```python
vector_row = vector.reshape(1, 10)
vector_t = vector.T
print('Original vector:\n', vector_row)
print("Транспонований vector \n", vector_t)
```

Так. Вектор був рядком, після транспонування має вигляд стовпця

3.

```python
column_vector = np.expand_dims(vector, axis=1)

print("Оригінальний vector\n",vector.shape,"\n", vector)
print("Стовпчиковий vector\n", column_vector.shape,"\n", column_vector)
```

vector має (10,)-це одновимірний вектор.

column_vector має(10, 1)-це двовимірний вектор.

```python
column_vector_t = column_vector.T
print('Original:\n', column_vector.shape,"\n", column_vector)
print("Транспонований: \n", column_vector_t.shape,"\n", column_vector_t)
```
