1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sp.Symbol('x')
```

```python
f = 10*x-5
```

```python
derivative_f_lin = sp.diff(f, x)
print('y=',derivative_f_lin)
```

```python
# Визначення функції та її похідної
x = np.linspace(0, 10, 200)
f = 10*x-5
df = np.full_like(x, 10)  # Похідна f'(x) = 10

# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = 10*x-5', color='blue')
plt.plot(x, df, label="f'(x) = 10", color='green')

# Позначення точки та дотичної лінії
x_point = 10
f_point = x_point*10-5
df_point = 10
tangent_line = df_point * (x - x_point) + f_point  # Рівняння дотичної: y - y1 = m(x - x1)

plt.scatter(x_point, f_point, color='red') # точка на кривій
plt.plot(x, tangent_line, '--', color='red', label='Tangent at x = 10') # дотична

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

Висновок:

Кожна додаткова година додає 10 балів.

Ефективність навчання не змінюється з часом, так як похідна стала величина.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.Symbol('x')
f = sp.sin(x)
derivative_f_lin = sp.diff(f, x)
print('y=',derivative_f_lin)
```

```python
# Визначення функції та її похідної
x = np.linspace(0*np.pi, 4*np.pi, 200)
f = np.sin(x)
df = np.cos(x)  # Похідна f'(x) = cos(x)

# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = sin(x)', color='blue')
plt.plot(x, df, label="f'(x) = cos(x)", color='green')

# Позначення точки та дотичної лінії
x_point = np.pi
f_point = np.sin(x_point)
df_point = np.cos(x_point)
tangent_line = df_point * (x - x_point) + f_point  # Рівняння дотичної: y - y1 = m(x - x1)

plt.scatter(x_point, f_point, color='red') # точка на кривій
plt.plot(x, tangent_line, '--', color='red', label='Tangent at x = pi') # дотична

xticks = [0, np.pi, 2*np.pi, 3*np.pi, 4*np.pi]
xtick_labels = ['0', 'π', '2π', '3π', '4π']
plt.xticks(xticks, xtick_labels)

# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
x = sp.Symbol('x')
f = sp.ln(x)
derivative_f_lin = sp.diff(f, x)
print('y=',derivative_f_lin)
```

```python
# Визначення функції та її похідної
x = np.linspace(0.1, np.exp(3), 200)
f = np.log(x)
df = 1/x  # Похідна f'(x) = 1/x

# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = log(x)', color='blue')
plt.plot(x, df, label="f'(x) = 1/x", color='green')

# Позначення точки та дотичної лінії
x_point = np.exp(1)
f_point = np.log(x_point)
df_point = 1/x_point
tangent_line = df_point * (x - x_point) + f_point  # Рівняння дотичної: y - y1 = m(x - x1)

plt.scatter(x_point, f_point, color='red') # точка на кривій
plt.plot(x, tangent_line, '--', color='red', label='Tangent at x = e') # дотична

xticks = [0.1, np.exp(1), np.exp(2), np.exp(3)]
xtick_labels = ['0.1', 'e', 'e^2', 'e^3']
plt.xticks(xticks, xtick_labels)


# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```
